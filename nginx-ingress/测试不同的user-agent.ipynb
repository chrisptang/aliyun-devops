```python
import requests

for i in range(0,10):
    unique_hostname=set()
    for run in range(0,5):
        # 每个User Agent都跑5遍
        user_agent=f"UA-{i}"
        # print(user_agent, f"http://localhost?run={i}_{run}")
        hostname=requests.get(f"http://localhost?run={i}_{run}", headers={"User-Agent": user_agent}).json().get("hostname")
        ua=requests.get(f"http://localhost?run={i}_{run}", headers={"User-Agent": user_agent}).json().get("headers").get("User-Agent")
        print(f"UA:{user_agent}, hostname:{hostname}, ua:{ua}")
        unique_hostname.add(hostname)
    print(f"{len(unique_hostname)} hosts funds.\n\n")
```
